```python
def chunk_text(texto, chunk_size=60, overlap=15):
    """Divide um texto em chunks com sobreposição."""
    palavras = texto.split()
    chunks = []
    inicio = 0
    while inicio < len(palavras):
        fim = inicio + chunk_size
        chunk = ' '.join(palavras[inicio:fim])
        chunks.append(chunk)
        inicio += chunk_size - overlap
    return chunks

# Gerar chunks com metadados
chunks_com_metadados = []
for item in corpus_raw:
    sub_chunks = chunk_text(item['texto'])
    for i, chunk in enumerate(sub_chunks):
        chunks_com_metadados.append({
            'id': f"{item['artigo'].replace(' ', '_').replace('º', '')}_{i}",
            'texto': chunk,
            'artigo': item['artigo'],
            'capitulo': item['capitulo']
        })

print(f'✅ Total de chunks gerados: {len(chunks_com_metadados)}')
print(f'\n📌 Exemplo de chunk com metadados:')
print(f"  ID      : {chunks_com_metadados[0]['id']}")
print(f"  Artigo  : {chunks_com_metadados[0]['artigo']}")
print(f"  Capítulo: {chunks_com_metadados[0]['capitulo']}")
print(f"  Texto   : {chunks_com_metadados[0]['texto']}")
```

```python
import chromadb
from sentence_transformers import SentenceTransformer

# Modelo multilíngue — funciona bem com português
print('🔄 Carregando modelo de embeddings...')
modelo_embedding = SentenceTransformer('paraphrase-multilingual-MiniLM-L12-v2')
print('✅ Modelo carregado!')

# Inicializar ChromaDB em memória
client = chromadb.Client()

# Criar coleção
try:
    client.delete_collection('cnj_615_2025')
except:
    pass

colecao = client.create_collection('cnj_615_2025')

# Gerar embeddings e indexar
textos = [c['texto'] for c in chunks_com_metadados]
ids = [c['id'] for c in chunks_com_metadados]
metadados = [{'artigo': c['artigo'], 'capitulo': c['capitulo']} for c in chunks_com_metadados]

print('🔄 Gerando embeddings e indexando no ChromaDB...')
embeddings = modelo_embedding.encode(textos).tolist()

colecao.add(
    documents=textos,
    embeddings=embeddings,
    ids=ids,
    metadatas=metadados
)

print(f'✅ {len(textos)} chunks indexados com sucesso!')
```

```python
def recuperar_chunks(pergunta, top_k=6):
    """Recupera os top-k chunks mais relevantes para a pergunta."""
    embedding_pergunta = modelo_embedding.encode([pergunta]).tolist()
    resultados = colecao.query(
        query_embeddings=embedding_pergunta,
        n_results=top_k,
        include=['documents', 'metadatas', 'distances']
    )
    chunks_recuperados = []
    for i in range(len(resultados['documents'][0])):
        chunks_recuperados.append({
            'texto': resultados['documents'][0][i],
            'artigo': resultados['metadatas'][0][i]['artigo'],
            'capitulo': resultados['metadatas'][0][i]['capitulo'],
            'distancia': resultados['distances'][0][i]
        })
    return chunks_recuperados

# Teste de recuperação
pergunta_teste = 'Quais são as vedações ao uso de IA no judiciário?'
chunks_recuperados = recuperar_chunks(pergunta_teste, top_k=6)

print(f'🔍 Pergunta: {pergunta_teste}')
print(f'\n📦 Top-{len(chunks_recuperados)} chunks recuperados (antes do reranking):\n')
for i, c in enumerate(chunks_recuperados):
    print(f'  [{i+1}] Distância: {c["distancia"]:.4f} | {c["artigo"]}')
    print(f'       {c["texto"][:100]}...')
    print()
```

```python
import numpy as np

def similaridade_cosseno(vec1, vec2):
    """Calcula a similaridade de cosseno entre dois vetores."""
    vec1, vec2 = np.array(vec1), np.array(vec2)
    return np.dot(vec1, vec2) / (np.linalg.norm(vec1) * np.linalg.norm(vec2))

def reranquear(pergunta, chunks, top_n=3):
    """Reordena os chunks por similaridade de cosseno e retorna os top-n."""
    embedding_pergunta = modelo_embedding.encode([pergunta])[0]

    for chunk in chunks:
        embedding_chunk = modelo_embedding.encode([chunk['texto']])[0]
        chunk['score_reranking'] = similaridade_cosseno(embedding_pergunta, embedding_chunk)

    chunks_reranqueados = sorted(chunks, key=lambda x: x['score_reranking'], reverse=True)
    return chunks_reranqueados[:top_n]

# Aplicar reranking
chunks_finais = reranquear(pergunta_teste, chunks_recuperados, top_n=3)

print(f'✅ Top-{len(chunks_finais)} chunks após re-ranqueamento:\n')
for i, c in enumerate(chunks_finais):
    print(f'  [{i+1}] Score: {c["score_reranking"]:.4f} | {c["artigo"]}')
    print(f'       {c["texto"][:120]}...')
    print()
```

```python
import requests

def gerar_resposta_maritaca(pergunta, chunks_contexto):
    """Gera resposta usando a API da Maritaca com os chunks reranqueados como contexto."""

    contexto = '\n\n'.join([
        f"[{c['artigo']} — {c['capitulo']}]\n{c['texto']}"
        for c in chunks_contexto
    ])

    prompt = f"""Você é um assistente jurídico especializado na Resolução CNJ nº 615/2025,
que estabelece diretrizes para uso de Inteligência Artificial no Poder Judiciário brasileiro.

Use APENAS o contexto abaixo para responder. Se a resposta não estiver no contexto, diga que
não encontrou a informação.

CONTEXTO:
{contexto}

PERGUNTA: {pergunta}

RESPOSTA:"""

    headers = {
        'Authorization': f'Key {MARITACA_API_KEY}',
        'Content-Type': 'application/json'
    }

    payload = {
        'model': 'sabia-3',
        'messages': [{'role': 'user', 'content': prompt}],
        'max_tokens': 500,
        'temperature': 0.3
    }

    response = requests.post(
        'https://chat.maritaca.ai/api/chat/completions',
        headers=headers,
        json=payload
    )

    if response.status_code == 200:
        return response.json()['choices'][0]['message']['content']
    else:
        return f'Erro na API: {response.status_code} — {response.text}'

# Gerar resposta
print(f'❓ Pergunta: {pergunta_teste}\n')
resposta = gerar_resposta_maritaca(pergunta_teste, chunks_finais)
print(f'💬 Resposta do sabia-3:\n')
print(resposta)
```

```python
def rag_avancado(pergunta, top_k=10, top_n=3):
    """
    Pipeline RAG Avançado completo:
    1. Recuperação Top-K no ChromaDB
    2. Re-ranqueamento por similaridade de cosseno
    3. Geração com Maritaca sabia-3
    """
    print(f'❓ Pergunta: {pergunta}')
    print('─' * 60)

    # Etapa 1: Recuperação
    chunks_recuperados = recuperar_chunks(pergunta, top_k=top_k)
    print(f'📦 {len(chunks_recuperados)} chunks recuperados (Top-{top_k})')

    # Etapa 2: Re-ranqueamento
    chunks_finais = reranquear(pergunta, chunks_recuperados, top_n=top_n)
    print(f'🔄 Re-ranqueamento aplicado → Top-{top_n} selecionados')
    print('\n📌 Chunks usados como contexto:')
    for i, c in enumerate(chunks_finais):
        print(f'  [{i+1}] Score: {c["score_reranking"]:.4f} | {c["artigo"]}')

    # Etapa 3: Geração
    print('\n💬 Resposta gerada pelo sabia-3:')
    print('─' * 60)
    resposta = gerar_resposta_maritaca(pergunta, chunks_finais)
    print(resposta)
    print('─' * 60)
    return resposta

# Testando com a mesma pergunta
rag_avancado('Quais são as vedações ao uso de IA no judiciário?')
```

```python
perguntas = [
    'Quais são as vedações ao uso de IA no judiciário?',
    'O que é proibido no desenvolvimento de IA nos tribunais?',
    'Quais são os fundamentos do uso de IA no Poder Judiciário?',
    'O que é explicabilidade segundo a Resolução CNJ 615/2025?',
    'Como os tribunais devem tratar dados usados no treinamento de IA?'
]

for p in perguntas:
    rag_avancado(p)
    print('\n')
```

```python
pergunta_comparativo = 'Como os tribunais devem tratar dados usados no treinamento de IA?'

# Antes: recuperação pura
chunks_sem = recuperar_chunks(pergunta_comparativo, top_k=6)

# Depois: reranking
chunks_com = reranquear(pergunta_comparativo, chunks_sem.copy(), top_n=3)

print('=' * 70)
print('📊 COMPARATIVO: Antes vs. Depois do Re-ranqueamento')
print('=' * 70)

print('\n🔴 ANTES — Top-6 por distância vetorial (ChromaDB):')
for i, c in enumerate(chunks_sem):
    print(f'  [{i+1}] Distância: {c["distancia"]:.4f} | {c["artigo"]}')

print('\n🟢 DEPOIS — Top-3 após re-ranqueamento por cosseno:')
for i, c in enumerate(chunks_com):
    print(f'  [{i+1}] Score: {c["score_reranking"]:.4f} | {c["artigo"]}')

print('\n✅ O re-ranqueamento reorganiza e filtra os chunks mais relevantes!')
```
